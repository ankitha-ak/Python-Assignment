```python
#Question 3
def isMatch(s, p):
    dp = [[False] * (len(p) + 1) for _ in range(len(s) + 1)]

    dp[0][0] = True

    for j in range(2, len(p) + 1):
        if p[j - 1] == '*':
            dp[0][j] = dp[0][j - 2]

    for i in range(1, len(s) + 1):
        for j in range(1, len(p) + 1):

            if p[j - 1] == '.' or p[j - 1] == s[i - 1]:
                dp[i][j] = dp[i - 1][j - 1]

            elif p[j - 1] == '*':
                dp[i][j] = dp[i][j - 2]

                if p[j - 2] == '.' or p[j - 2] == s[i - 1]:
                    dp[i][j] = dp[i][j] or dp[i - 1][j]

    return dp[len(s)][len(p)]


print(isMatch("abcd", ".*"))
```

```python
#Question 4
def decode(s):
    result = []

    def backtrack(index, current):
        if index == len(s):
            result.append(current)
            return

        if s[index] == '0':
            return

        num = int(s[index])
        backtrack(index + 1, current + chr(64 + num))

        if index + 1 < len(s):
            num = int(s[index:index + 2])

            if 10 <= num <= 26:
                backtrack(index + 2, current + chr(64 + num))

    backtrack(0, "")

    print("Total Ways:", len(result))
    print("Decoded Strings:")

    for item in result:
        print(item)


decode("226")
```

```python
#Question 5
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None


def isBalanced(root):

    def height(node):
        if node is None:
            return 0

        left = height(node.left)
        right = height(node.right)

        if left == -1 or right == -1:
            return -1

        if abs(left - right) > 1:
            return -1

        return max(left, right) + 1

    return height(root) != -1


root = Node(3)
root.left = Node(9)
root.right = Node(20)
root.right.left = Node(15)
root.right.right = Node(7)

print(isBalanced(root))
```

```python
#Question 6
def reverseInteger(x):
    sign = -1 if x < 0 else 1
    x = abs(x)

    rev = 0

    while x > 0:
        digit = x % 10
        rev = rev * 10 + digit
        x //= 10

    rev *= sign

    if rev < -2**31 or rev > 2**31 - 1:
        return 0

    return rev


print(reverseInteger(123))
print(reverseInteger(-123))
```

```python
#Question 7
from itertools import combinations

def solve(L, K):
    strings = []

    for positions in combinations(range(L), K):
        s = ['0'] * L

        for p in positions:
            s[p] = '1'

        strings.append(''.join(s))

    minimum = float('inf')
    results = []

    for s in strings:
        blocks = s.split('1')
        maximum = max(map(len, blocks))

        print(s, "-> max zeros =", maximum)

        if maximum < minimum:
            minimum = maximum
            results = [s]
        elif maximum == minimum:
            results.append(s)

    print("Optimal Arrangements:", results)
    print("Minimum Longest Zero Block:", minimum)


solve(4, 2)
```

```python
#Question 8
def mergeTriplets(triplets, target):

    found = [False, False, False]

    for a, b, c in triplets:

        if a <= target[0] and b <= target[1] and c <= target[2]:

            if a == target[0]:
                found[0] = True

            if b == target[1]:
                found[1] = True

            if c == target[2]:
                found[2] = True

    return all(found)


triplets = [[2,5,3], [1,8,4], [1,7,5]]
target = [2,7,5]

print(mergeTriplets(triplets, target))
```

```python
#Question 12
def coinChange(coins, amount):

    dp = [float('inf')] * (amount + 1)

    dp[0] = 0

    for i in range(1, amount + 1):
        for coin in coins:

            if coin <= i:
                dp[i] = min(dp[i], dp[i - coin] + 1)

    if dp[amount] == float('inf'):
        return -1

    return dp[amount]


coins = [1, 2, 5]
amount = 11

print(coinChange(coins, amount))
```

```python
#Question 13
def findCheapestPrice(n, flights, src, dst, k):

    price = [float('inf')] * n
    price[src] = 0

    for i in range(k + 1):

        temp = price.copy()

        for u, v, cost in flights:

            if price[u] != float('inf'):
                temp[v] = min(temp[v], price[u] + cost)

        price = temp

    if price[dst] == float('inf'):
        return -1

    return price[dst]


n = 4

flights = [
    [0,1,100],
    [1,2,100],
    [2,0,100],
    [1,3,600],
    [2,3,200]
]

print(findCheapestPrice(n, flights, 0, 3, 1))
```

```python
#Question 14
import heapq

def mergeKLists(lists):

    result = []

    for lst in lists:
        for num in lst:
            heapq.heappush(result, num)

    output = []

    while result:
        output.append(heapq.heappop(result))

    return output


lists = [[1,4,5], [1,3,4], [2,6]]

print(mergeKLists(lists))
```

```python
#Question 17
def merge(intervals):
    intervals.sort()
    result = []

    for start, end in intervals:
        if not result or start > result[-1][1]:
            result.append([start, end])
        else:
            result[-1][1] = max(result[-1][1], end)

    return result


print(merge([[1, 3], [2, 6], [8, 10], [15, 18]]))
```

```python
#Question 18
def max_product(nums):
    maximum = nums[0]
    minimum = nums[0]
    result = nums[0]

    for num in nums[1:]:
        if num < 0:
            maximum, minimum = minimum, maximum

        maximum = max(num, maximum * num)
        minimum = min(num, minimum * num)

        result = max(result, maximum)

    return result


print(max_product([2, 3, -2, 4]))
```

```python
#Question 19
def smallest_team(required, people):
    n = len(required)
    target = (1 << n) - 1

    dp = {0: []}

    for i, person in enumerate(people):
        mask = 0

        for skill in person:
            if skill in required:
                mask |= 1 << required.index(skill)

        for old_mask, team in list(dp.items()):
            new_mask = old_mask | mask

            if new_mask not in dp or len(dp[new_mask]) > len(team) + 1:
                dp[new_mask] = team + [i]

    return dp[target]


required = ["Python", "SQL", "MachineLearning", "DataStructures"]

people = [
    ["Python", "SQL"],
    ["SQL", "MachineLearning"],
    ["MachineLearning", "DataStructures"],
    ["DataStructures"]
]

print(smallest_team(required, people))
```

```python
#Question 20
def word_break(s, wordDict):
    word_set = set(wordDict)
    memo = {}

    def solve(start):
        if start == len(s):
            return [""]

        if start in memo:
            return memo[start]

        result = []

        for end in range(start + 1, len(s) + 1):
            word = s[start:end]

            if word in word_set:
                for rest in solve(end):
                    result.append(word + (" " + rest if rest else ""))

        memo[start] = result
        return result

    patterns = solve(0)

    print("Can Segment:", bool(patterns))
    print("Segmentations:")

    for pattern in patterns:
        print(pattern)

    if not patterns:
        print("None")


word_break("catsanddog", ["cat", "cats", "and", "sand", "dog"])
```

```python
#Question 24
import math

def closest_pair(points):
    minimum = float('inf')

    for i in range(len(points)):
        for j in range(i + 1, len(points)):
            x1, y1 = points[i]
            x2, y2 = points[j]

            distance = math.sqrt(
                (x2 - x1) ** 2 + (y2 - y1) ** 2
            )

            minimum = min(minimum, distance)

    return minimum


points = [(0, 0), (1, 1)]

print(closest_pair(points))
```

```python
#Question 29
def is_happy(n):
    seen = set()

    while n != 1 and n not in seen:
        seen.add(n)

        total = 0

        while n > 0:
            digit = n % 10
            total += digit ** 2
            n //= 10

        n = total

    return n == 1


print(is_happy(19))
print(is_happy(2))
```

```python
#Question 30
def set_zeroes(matrix):
    rows = set()
    cols = set()

    for i in range(len(matrix)):
        for j in range(len(matrix[0])):
            if matrix[i][j] == 0:
                rows.add(i)
                cols.add(j)

    for i in range(len(matrix)):
        for j in range(len(matrix[0])):
            if i in rows or j in cols:
                matrix[i][j] = 0

    return matrix


matrix = [
    [1, 1, 1],
    [1, 0, 1],
    [1, 1, 1]
]

print(set_zeroes(matrix))
```

```python
#Question 31
def card_probability(cards):
    total = len(cards)

    ace = 0
    face = 0
    red = 0
    black = 0

    for card in cards:
        rank = card[:-1]
        suit = card[-1]

        if rank == "A":
            ace += 1

        if rank in ["J", "Q", "K"]:
            face += 1

        if suit in ["H", "D"]:
            red += 1
        else:
            black += 1

    print("P(Ace) =", ace / total)
    print("P(Face) =", face / total)
    print("P(Red) =", red / total)

    categories = {
        "Ace": ace,
        "Face": face,
        "Red": red
    }

    print("Most Frequent Category:", max(categories, key=categories.get))
    print("P(Black) =", black / total)


cards = ["AH", "10D", "KC", "QS", "2H", "JD", "7C", "KH"]

card_probability(cards)
```

```python
#Question 35
def trap(height):
    left = 0
    right = len(height) - 1

    left_max = 0
    right_max = 0
    water = 0

    while left < right:

        if height[left] < height[right]:

            if height[left] >= left_max:
                left_max = height[left]
            else:
                water += left_max - height[left]

            left += 1

        else:

            if height[right] >= right_max:
                right_max = height[right]
            else:
                water += right_max - height[right]

            right -= 1

    return water


print(trap([0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]))
print(trap([4, 2, 0, 3, 2, 5]))
```

```python
#Question 36
def generate_parentheses(n):
    result = []

    def backtrack(current, open_count, close_count):

        if len(current) == 2 * n:
            result.append(current)
            return

        if open_count < n:
            backtrack(current + "(", open_count + 1, close_count)

        if close_count < open_count:
            backtrack(current + ")", open_count, close_count + 1)

    backtrack("", 0, 0)

    return result


print(generate_parentheses(3))
print(generate_parentheses(1))
```

```python
#Question 37
def spiral_order(matrix):
    result = []

    top = 0
    bottom = len(matrix) - 1
    left = 0
    right = len(matrix[0]) - 1

    while top <= bottom and left <= right:

        for i in range(left, right + 1):
            result.append(matrix[top][i])
        top += 1

        for i in range(top, bottom + 1):
            result.append(matrix[i][right])
        right -= 1

        if top <= bottom:
            for i in range(right, left - 1, -1):
                result.append(matrix[bottom][i])
            bottom -= 1

        if left <= right:
            for i in range(bottom, top - 1, -1):
                result.append(matrix[i][left])
            left += 1

    return result


matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(spiral_order(matrix))
```

```python
#Question 38
def max_area(height):
    left = 0
    right = len(height) - 1

    maximum = 0

    while left < right:

        width = right - left
        h = min(height[left], height[right])

        area = width * h

        maximum = max(maximum, area)

        if height[left] < height[right]:
            left += 1
        else:
            right -= 1

    return maximum


print(max_area([1, 8, 6, 2, 5, 4, 8, 3, 7]))
print(max_area([1, 1]))
print(max_area([4, 3, 2, 1, 4]))
```

```python
#Question 39
def matrix_chain(p):
    n = len(p) - 1

    dp = [[0] * n for _ in range(n)]
    split = [[0] * n for _ in range(n)]

    for length in range(2, n + 1):

        for i in range(n - length + 1):
            j = i + length - 1
            dp[i][j] = float('inf')

            for k in range(i, j):

                cost = (dp[i][k] + dp[k + 1][j]
                        + p[i] * p[k + 1] * p[j + 1])

                if cost < dp[i][j]:
                    dp[i][j] = cost
                    split[i][j] = k

    def print_order(i, j):
        if i == j:
            return "A" + str(i + 1)

        k = split[i][j]

        return "(" + print_order(i, k) + " x " + print_order(k + 1, j) + ")"

    print("Minimum Cost:", dp[0][n - 1])
    print("Order:", print_order(0, n - 1))


matrix_chain([10, 30, 5, 60])
```

```python
#Question 40
def solve_sudoku(board):

    for row in range(9):
        for col in range(9):

            if board[row][col] == '.':

                for num in "123456789":

                    if is_valid(board, row, col, num):
                        board[row][col] = num

                        if solve_sudoku(board):
                            return True

                        board[row][col] = '.'

                return False

    return True


def is_valid(board, row, col, num):

    for i in range(9):
        if board[row][i] == num:
            return False

        if board[i][col] == num:
            return False

    start_row = (row // 3) * 3
    start_col = (col // 3) * 3

    for i in range(start_row, start_row + 3):
        for j in range(start_col, start_col + 3):
            if board[i][j] == num:
                return False

    return True


board = [
    list("53..7...."),
    list("6..195..."),
    list(".98....6."),
    list("8...6...3"),
    list("4..8.3..1"),
    list("7...2...6"),
    list(".6....28."),
    list("...419..5"),
    list("....8..79")
]

if solve_sudoku(board):
    for row in board:
        print("".join(row))
else:
    print("No solution")
```

```python
#Question 41
def find_words(board, words):
    result = []

    rows = len(board)
    cols = len(board[0])

    def dfs(r, c, word, index):

        if index == len(word):
            return True

        if (r < 0 or r >= rows or
            c < 0 or c >= cols or
            board[r][c] != word[index]):
            return False

        temp = board[r][c]
        board[r][c] = '#'

        found = (
            dfs(r + 1, c, word, index + 1) or
            dfs(r - 1, c, word, index + 1) or
            dfs(r, c + 1, word, index + 1) or
            dfs(r, c - 1, word, index + 1)
        )

        board[r][c] = temp

        return found

    for word in words:
        found = False

        for r in range(rows):
            for c in range(cols):
                if dfs(r, c, word, 0):
                    found = True
                    break
            if found:
                break

        if found:
            result.append(word)

    return result


board = [
    ["o", "a", "a", "n"],
    ["e", "t", "a", "e"],
    ["i", "h", "k", "r"],
    ["i", "f", "l", "v"]
]

words = ["oath", "pea", "eat", "rain"]

print(find_words(board, words))
```

```python
#Question 43
from collections import defaultdict

def euler_path(n, edges):

    graph = defaultdict(list)
    degree = [0] * n

    for u, v in edges:
        graph[u].append(v)
        graph[v].append(u)

        degree[u] += 1
        degree[v] += 1

    odd = [i for i in range(n) if degree[i] % 2 == 1]

    if len(odd) not in [0, 2]:
        return False

    if not edges:
        return []

    start = odd[0] if odd else next(i for i in range(n) if degree[i] > 0)

    visited = set()
    stack = [start]

    while stack:
        node = stack.pop()

        if node not in visited:
            visited.add(node)

            for neighbour in graph[node]:
                stack.append(neighbour)

    for i in range(n):
        if degree[i] > 0 and i not in visited:
            return False

    graph = defaultdict(list)

    for u, v in edges:
        graph[u].append(v)
        graph[v].append(u)

    used = set()
    stack = [start]
    path = []

    while stack:
        node = stack[-1]

        while graph[node] and tuple(sorted((node, graph[node][-1]))) in used:
            graph[node].pop()

        if graph[node]:
            neighbour = graph[node].pop()
            edge = tuple(sorted((node, neighbour)))

            used.add(edge)
            stack.append(neighbour)
        else:
            path.append(stack.pop())

    path.reverse()

    if len(used) != len(edges):
        return False

    return path


edges = [[0, 1], [1, 2], [2, 3]]

print(euler_path(4, edges))
```

```python
#Question 44
def rotate(matrix):
    n = len(matrix)

    # Transpose the matrix
    for i in range(n):
        for j in range(i + 1, n):
            matrix[i][j], matrix[j][i] = (
                matrix[j][i], matrix[i][j]
            )

    # Reverse each row
    for row in matrix:
        row.reverse()

    return matrix


matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(rotate(matrix))
```

```python
#Question 45
def climbing_stairs(n):
    result = []

    def generate(remaining, path):

        if remaining == 0:
            result.append(path)
            return

        if remaining >= 1:
            generate(remaining - 1, path + [1])

        if remaining >= 2:
            generate(remaining - 2, path + [2])

    generate(n, [])

    print("Total Ways:", len(result))
    print("Patterns:")

    for pattern in result:
        print("+".join(map(str, pattern)))


climbing_stairs(3)
```

```python
#Question 52
import csv
import json
from io import StringIO

raw_data = """Name,Age,Salary,Join Date
Alice,30,95000,2015-03-01
Bob,,60000,2018-07-15
Charlie,35,Unknown,2024-02-01
Diana,40,110000,2021-11-03"""

cleaned_records = []

reader = csv.DictReader(StringIO(raw_data))

for row in reader:
    try:
        name = row["Name"].strip()
        age = row["Age"].strip()
        salary = row["Salary"].strip()
        join_date = row["Join Date"].strip()

        if not age or not salary or salary == "Unknown":
            continue

        record = {
            "name": name,
            "age": int(age),
            "salary": float(salary),
            "join_date": join_date
        }

        cleaned_records.append(record)

    except ValueError:
        continue

with open("cleaned_data.json", "w") as file:
    json.dump(cleaned_records, file, indent=2)

print(json.dumps(cleaned_records, indent=2))
```

```python
#Question 53
def longest_substring(s):
    visited = [False] * 26

    left = 0
    maximum = 0

    for right in range(len(s)):
        index = ord(s[right]) - ord('a')

        while visited[index]:
            old_index = ord(s[left]) - ord('a')
            visited[old_index] = False
            left += 1

        visited[index] = True

        maximum = max(maximum, right - left + 1)

    return maximum

s = "abcabcbb"

print("Length of longest unique substring:", longest_substring(s))
```

```python
#Question 55
class NumberIterator:

    def __init__(self, start, stop, step):
        self.current = start
        self.stop = stop
        self.step = step

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.stop + 0.000001:
            raise StopIteration

        value = round(self.current, 2)
        self.current += self.step

        return value

numbers = NumberIterator(0.0, 1.0, 0.25)

for num in numbers:
    print(num)
```

```python
#Question 57
from collections import Counter
from functools import reduce
from datetime import date
import random

students = {
    "Tarun": [82, 91, 78],
    "Karthik": [65, 72, 68],
    "Chadrika": [91, 95, 89],
    "Trisha": [76, 81, 79],
    "Atul": [88, 84, 92]
}

averages = {}

for name, marks in students.items():
    averages[name] = sum(marks) / len(marks)

print("Student Averages:")

for name, avg in averages.items():
    print(name, ":", round(avg, 2))

highest = max(averages, key=averages.get)
print("Highest Average:", highest)

selected = random.choice(list(students.keys()))
print("Random Student:", selected)

print("Current Date:", date.today())

grades = []

for avg in averages.values():
    if avg >= 90:
        grades.append("A")
    elif avg >= 80:
        grades.append("B")
    elif avg >= 70:
        grades.append("C")
    elif avg >= 60:
        grades.append("D")
    else:
        grades.append("F")

print("Grade Count:", Counter(grades))

total = reduce(
    lambda a, b: a + b,
    [mark for marks in students.values() for mark in marks]
)

print("Total Marks:", total)
```

```python
#Question 58
from collections import defaultdict
from itertools import combinations
from random import choice
from functools import reduce
from datetime import date
import math
import sys

students = [
    ("Anita", "CSE", 85),
    ("Rahul", "CSE", 78),
    ("Priya", "CSE", 92),
    ("Kiran", "CSE", 88),
    ("Asha", "AI", 80),
    ("Ravi", "AI", 75),
    ("Meena", "AI", 90),
    ("Arun", "AI", 82),
    ("John", "Data Science", 86),
    ("Sara", "Data Science", 91),
    ("Vijay", "Cyber Security", 79),
    ("Neha", "Cyber Security", 95)
]

try:
    team_size = int(sys.argv[1])
except (IndexError, ValueError):
    team_size = 3

groups = defaultdict(list)

for student in students:
    groups[student[1]].append(student)

department = "CSE"

possible_teams = list(combinations(groups[department], team_size))

if not possible_teams:
    print("No team available")
else:
    team = choice(possible_teams)

    marks = [student[2] for student in team]

    total = reduce(lambda a, b: a + b, marks)
    average = total / len(team)

    print("===== PROJECT TEAM ALLOCATION =====")
    print("Team Size:", team_size)
    print("Allocation Date:", date.today())
    print("Department:", department)
    print("Selected Team:")

    for i, student in enumerate(team, 1):
        print(i, student[0], "-", student[2])

    print("Total Marks:", total)
    print("Average Marks:", math.ceil(average * 100) / 100)
```

```python
#Question 59
class InvalidMarksError(Exception):
    pass

while True:
    try:
        name = input("Enter student name: ")

        assert name.strip() != "", "Name cannot be empty"

        marks = float(input("Enter marks: "))

        if marks < 0 or marks > 100:
            raise InvalidMarksError("Marks must be between 0 and 100")

        with open("student_records.txt", "a") as file:
            file.write(name + " " + str(marks) + "\n")

        print("Record saved successfully")

    except AssertionError as e:
        print(e)

    except ValueError:
        print("Please enter numeric marks")

    except InvalidMarksError as e:
        print(e)

    finally:
        print("Record processing completed")

    choice = input("Do you want to continue? (yes/no): ")

    if choice.lower() != "yes":
        break
```

```python
#Question 61
def calculate_interest(metal, purity, amount, tenure):

    rates = {
        "gold": {18: 8, 22: 7, 24: 6},
        "silver": {925: 9, 950: 8, 999: 7},
        "platinum": {900: 10, 950: 9, 999: 8}
    }

    metal = metal.lower()

    if metal not in rates:
        raise ValueError("Invalid metal type")

    if purity not in rates[metal]:
        raise ValueError("Invalid purity")

    rate = rates[metal][purity]

    interest = amount * rate * tenure / 100

    return interest


try:
    result = calculate_interest("gold", 22, 100000, 2)

except TypeError as e:
    print("Missing argument:", e)

except ValueError as e:
    print("Error:", e)

else:
    print("Interest Amount:", result)

finally:
    print("Loan processing completed")
```

```python
#Question 62
def solve_maze(maze):

    n = len(maze)

    solution = [[0] * n for _ in range(n)]

    def solve(x, y):

        if x == n - 1 and y == n - 1:
            solution[x][y] = 1
            return True

        if (x < 0 or y < 0 or x >= n or y >= n
                or maze[x][y] == 0 or solution[x][y] == 1):
            return False

        solution[x][y] = 1

        if (solve(x + 1, y) or
            solve(x, y + 1) or
            solve(x - 1, y) or
            solve(x, y - 1)):
            return True

        solution[x][y] = 0
        return False

    if maze[0][0] == 0 or maze[n-1][n-1] == 0:
        return None

    if solve(0, 0):
        return solution

    return None


maze = [
    [1, 0, 0, 0],
    [1, 1, 0, 1],
    [0, 1, 0, 0],
    [1, 1, 1, 1]
]

result = solve_maze(maze)

if result:
    print("Solution Path:")

    for row in result:
        print("".join(map(str, row)))
else:
    print("No solution exists")
```

```python
#Question 64
class Printer:

    def __init__(self, capacity, color, printer_type):
        self.capacity = capacity
        self.color = color
        self.printer_type = printer_type

    def display(self):
        print("Capacity:", self.capacity)
        print("Color:", self.color)
        print("Type:", self.printer_type)


printer1 = Printer(100, "Color", "Inkjet")
printer2 = Printer(200, "Mono", "Laser")
printer3 = Printer(150, "Color", "Laser")

print("Printer 1")
printer1.display()
print("Object ID:", id(printer1))

print("\nPrinter 2")
printer2.display()
print("Object ID:", id(printer2))

print("\nPrinter 3")
printer3.display()
print("Object ID:", id(printer3))
```
